## 5th Delivery Analytics Gold Dataset

```python
delivery_analytics_df = orders_silver_df.select(
    "order_id",
    "customer_id",
    "order_status",
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
)

print("Delivery Analytics rows:", delivery_analytics_df.count())
```

```python
from pyspark.sql.functions import datediff, to_date
```

```python
delivery_analytics_df = delivery_analytics_df.withColumn("delivery_days",
    when(col("order_delivered_customer_date").isNotNull(),datediff(
            to_date(col("order_delivered_customer_date")),
            to_date(col("order_purchase_timestamp"))
        )
    )
)

delivery_analytics_df.select(
    "order_id",
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "delivery_days"
).show(10)
```

```python
delivery_analytics_df = delivery_analytics_df.withColumn("delivery_delay_days",
    when(
        col("order_delivered_customer_date").isNotNull(),datediff(
            to_date(col("order_delivered_customer_date")),
            col("order_estimated_delivery_date")
        )
    )
)

delivery_analytics_df.select(
    "order_id",
    "order_estimated_delivery_date",
    "order_delivered_customer_date",
    "delivery_delay_days"
).show(10)
```

```python
delivery_analytics_df = delivery_analytics_df.withColumn("delivery_status",
    when(
        col("order_delivered_customer_date").isNull(),
        "not_delivered"
    ).when(
        col("delivery_delay_days") <= 0,
        "on_time"
    ).otherwise(
        "late"
    )
)

delivery_analytics_df.groupBy(
    "delivery_status"
).count().show()
```

```python
delivery_analytics_df = delivery_analytics_df.withColumn("carrier_handling_days",
    when(
        col("order_delivered_carrier_date").isNotNull(),
        datediff(
            to_date(col("order_delivered_carrier_date")),
            to_date(col("order_purchase_timestamp"))
        )
    )
)

delivery_analytics_df.select(
    "order_id",
    "order_purchase_timestamp",
    "order_delivered_carrier_date",
    "carrier_handling_days"
).show(10)
```

```python
delivery_analytics_df.select(
    sum(when(col("delivery_days") < 0, 1).otherwise(0)).alias("negative_delivery_days"),
    sum(when(col("carrier_handling_days") < 0, 1).otherwise(0)).alias("negative_carrier_days"),
    sum(when(col("delivery_status").isNull(), 1).otherwise(0)).alias("null_delivery_status")
).show()
```

```python
delivery_analytics_df = delivery_analytics_df.withColumn("source_system",lit("olist"))\
        .withColumn("processing_timestamp",current_timestamp())

delivery_analytics_df.printSchema()
```
